```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/prathyush/Downloads/iris.csv")
df['species'].unique()
```

```python
df['category_label'] = df['species'].map({'setosa':0,'versicolor':1,'virginica':2})
df
```

```python
df = df.iloc[np.random.permutation(len(df))]

df
```

```python
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import normalize
x=df.iloc[:,:4].values
y=df.iloc[:,5].values
y=to_categorical(y,num_classes=3)

xnormalized = normalize(x,axis=0)
xnormalized
```

```python
from sklearn.model_selection import train_test_split

xtrain,xtest,ytrain,ytest = train_test_split(xnormalized,y,test_size=0.2)
```

```python
from keras.models import Sequential
from keras.layers import Dense

model = Sequential()

model.add(Dense(32, input_dim=4, activation='relu')) #1st hidden layer
model.add(Dense(64, activation='relu')) #2nd hidden layer
model.add(Dense(128, activation='relu')) #3rd hidden layer
model.add(Dense(256, activation="relu"))


model.add(Dense(3, activation='softmax')) #output layer neurons depends on no.of categories(multi class classification)
```

```python
model.compile(
    loss='categorical_crossentropy', #error
    optimizer='adam', #gradient descent
    metrics=['accuracy'] #accuracy
)
```

```python
model.fit(xtrain,ytrain,
          validation_data=(xtest,ytest),
          batch_size=30,
          epochs=30)
```

```python
pred = model.predict(xtest)

y_label = np.argmax(ytest,axis=1)
pred_label = np.argmax(pred,axis=1)

accuracy=np.sum(y_label==pred_label)/len(pred) * 100
print(accuracy)
```
