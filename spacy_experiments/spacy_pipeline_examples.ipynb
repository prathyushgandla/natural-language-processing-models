```python
#Remove stopwords and punctuation, keep only meaningful tokens.
txt = "Wow! What a view, she exclaimed, it's beautiful."
doc = nlp(txt)
filtered_tokens=[]
for token in doc:
    if token.is_stop or token.is_punct:
        continue
    filtered_tokens.append(token.lemma_)
print(" ".join(filtered_tokens))
```

```python
#Link entities logically (Person → Org → Location)
txt = "John works at Google in California."
doc = nlp(txt)
required_tokens = {'PERSON':[],'ORG':[],'GPE':[]}
for i in doc.ents:
    if i.label_ in required_tokens:
        required_tokens[i.label_].append(i)
print(required_tokens)
```

```python
#Extract subject–verb–object (SVO) triplets
txt = "Alice gave Bob a book."
doc = nlp(txt)
subjects=[]
verbs=[]
objects=[]
for word in doc:
    if word.dep_ in ('nsubj', 'nsubjpass'): #token.dep_: This attribute provides the syntactic dependency label for the token.
        subjects.append(word)
    if word.pos_ == 'VERB':
        verbs.append(word)
    if word.dep_ in ('obj', 'dobj', 'pobj'): # object of preposition(pobj), direct of object(dobj)
        objects.append(word)
print(subjects,verbs,objects)
```

```python
#Find all verbs and their noun objects
text = "Bob ate pizza and drank coffee."
doc = nlp(text)
pairs = []
for token in doc:
    if token.pos_ == 'VERB':
        for child in token.children:
            if child.dep_ in ('dobj','obj'):
                pairs.append((token,child))
print(pairs)
```

```python
#Detect negation
text = "He does not like apples."
doc = nlp(text)

negated_verbs = []
for token in doc:
    if token.pos_ == 'VERB':
        negation = any(child.dep_ == 'neg' for child in token.children)
        if(negation):
            negated_verbs.append(token)
print(negated_verbs)
```

```python
#Find all adjective–noun pairs
text = "The big red car is fast."
doc = nlp(text)

pairs = []
for token in doc:
    if token.pos_ == 'NOUN':
        for child in token.children:
            if child.pos_ == 'ADJ':
                pairs.append((token,child))
print(pairs)
```

```python
#Extract job titles (e.g., “senior software engineer”)
from spacy.matcher import PhraseMatcher

matcher = PhraseMatcher(nlp.vocab)
job_titles = ["Data Scientist","Project Manager","Software Engineer"]
pattern = [nlp(title) for title in job_titles]
matcher.add('JOB_TITLES',pattern)

text = "Alice is a senior Software Engineer and Bob is a Data Scientist."

doc = nlp(text)
print(matcher(doc))
for match_id, start, end in matcher(doc):
    print(doc[start:end])
```

```python
#Extract product names using PhraseMatcher
products = ["iPhone", "Galaxy S10", "Pixel 8", "Xbox One"]
matcher = PhraseMatcher(nlp.vocab)
pattern = [nlp(product) for product in products]
matcher.add('PRODUCTS',pattern)

text = "I own an iPhone but my friend uses a Pixel 8. We also play Xbox One."
doc = nlp(text)

for match_id,start,end in matcher(doc):
    print(doc[start:end])
```

```python
#Normalize slang text
slang_map = {
    "u": "you",
    "r": "are",
    "gr8": "great",
    "lol": "laugh",
}
txt = "u r gr8!! lol"
doc = nlp(txt)
res = []
for token in doc:
    if token.text.lower() in slang_map:
        res.append(slang_map[token.text.lower()])
    else:
        res.append(token.text)
print(" ".join(res))
```

```python
#Conversation Analysis'
#Identify speaker
# Detect questions vs statements
# Find sentiment per sentence
from textblob import TextBlob

text = """
Alice: How are you doing today?
Bob: I'm doing great! Are we meeting later?
Alice: Yes, at 5 PM.
"""

lines = [l.strip() for l in text.split('\n') if l.strip()]

conversation=[]
for line in lines:
    if ':' in line:
        speaker, sentence = line.split(':')
        
        is_question = sentence.strip().endswith('?')
        sentiment = TextBlob(sentence).sentiment.polarity

        conversation.append({
            "speaker": speaker,
            "sentence": sentence.strip(),
            "is_question": is_question,
            "sentiment": sentiment
        })
for i in conversation:
    print(i)
```
